```python
# Problema:
# Antigüedad de un Auto Usado vs. Precio (x -> y)
# Variable X (Independiente): Años de uso o antigüedad del vehículo (en años).
# Variable Y (Dependiente): Precio del auto (en miles de dólares).
# Relación esperada: A mayor antigüedad, menor precio (pendiente b negativa).
# Por qué es interesante: Te permite probar tu algoritmo con una pendiente negativa (b < 0), 
# demostrando que la prueba y error funciona para cualquier tipo de inclinación.

# DataSet
anios_x = [
    0.5, 0.8, 1.2, 1.8, 2.3, 2.7, 3.1, 3.8, 4.2, 4.9,
    5.2, 5.7, 6.1, 6.8, 7.3, 7.9, 8.4, 8.8, 9.3, 10.0,
    10.5, 11.2, 11.8, 12.4, 13.0, 13.7, 14.2, 14.8, 15.3, 15.9,
    16.4, 17.0, 17.5, 18.1, 18.7, 19.2, 19.8, 20.3, 20.9, 21.4,
    22.0, 22.6, 23.1, 23.7, 24.2, 24.8, 25.1, 25.5, 25.8, 26.0
]

precio_y = [
    27.5, 24.0, 23.2, 22.0, 25.5, 19.2, 23.0, 15.5, 20.8, 18.0,
    21.5, 14.8, 23.0, 11.2, 17.0, 13.5, 19.8, 13.0, 15.2, 10.5,
    15.0, 9.2, 14.5, 7.8, 13.2, 6.5, 12.0, 5.8, 11.5, 4.2,
    10.8, 8.2, 7.5, 3.5, 8.8, 5.5, 3.8, 4.5, 9.0, 2.8,
    6.5, 1.8, 7.8, 3.2, 4.5, 2.2, 5.0, 3.5, 1.0, 3.0
]

print("=== DATASET DE AUTOS USADOS ===")
print("Años (X) | Precio en miles $ (Y)")
print("-" * 30)
for i in range(len(anios_x)):
    print(f"   {anios_x[i]}     |     ${precio_y[i]}k")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(anios_x, precio_y, color='blue', s=30, label='Datos Reales (Autos)')
plt.title('Relación entre Antigüedad y Precio del Auto', fontsize=12)
plt.xlabel('Antigüedad (Años)', fontsize=10)
plt.ylabel('Precio (Miles de $)', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.show()
```

```python
a_opciones = []
val_a = -100.0
while val_a <= 100.0:
    a_opciones.append(round(val_a, 2))
    val_a += 0.5

b_opciones = []
val_b = -100.0
while val_b <= 100.0:
    b_opciones.append(round(val_b, 2))
    val_b += 0.5

mejor_a = None
mejor_b = None
menor_error_promedio = float('inf')

for a_prueba in a_opciones:
    for b_prueba in b_opciones:
        
        error_total = 0
        n = len(anios_x)
        
        for i in range(n):
            x_real = anios_x[i]
            y_real = precio_y[i]
            
            y_calculada = a_prueba + (b_prueba * x_real)
            
            error_total += abs(y_real - y_calculada)
            
        error_promedio = error_total / n
        
        if error_promedio < menor_error_promedio:
            menor_error_promedio = error_promedio
            mejor_a = a_prueba
            mejor_b = b_prueba

print("RESULTADOS DEL ALGORITMO DE PRUEBA Y ERROR")
print(f"\nMejor valor de 'a': {mejor_a}")
print(f"Mejor valor de 'b' : {mejor_b}")
print(f"Error promedio por auto: {menor_error_promedio:.2f}")
print(f"\nEcuación obtenida: y = {mejor_a} + ({mejor_b}) * x")
```

```python
x_recta = list(range(0, 25))
y_recta = [mejor_a + (mejor_b * x) for x in x_recta]

plt.figure(figsize=(9, 6))

plt.scatter(anios_x, precio_y, color='blue', s=35, label='Autos Reales (Dataset)', zorder=5)
plt.plot(x_recta, y_recta, color='red', linestyle='-', linewidth=2, label=f'Recta Ajustada: y = {mejor_a} + ({mejor_b})x')

plt.title('Ajuste de Recta por Prueba y Error (Depreciación de Autos)', fontsize=12, fontweight='bold')
plt.xlabel('Antigüedad del Auto (Años)', fontsize=10)
plt.ylabel('Precio (Miles de $)', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')

plt.show()
```

```python
casos_prueba_x = [0.2, 1.5, 3.0, 5.5, 8.0, 10.5, 14.0, 18.5, 22.0, 24.3]

print("=== 10 PREDICCIONES DE PRECIO SEGÚN LA ANTIGÜEDAD DE UN AUTO ===")
print("  Antigüedad (X)   |   Precio Estimado (Y')")
print("-" * 48)

for x_nuevo in casos_prueba_x:

    y_prediccion = mejor_a + (mejor_b * x_nuevo)
    
    if y_prediccion < 0:
        y_prediccion = 0.0
        
    print(f"  Auto de {x_nuevo:<4} años   |   ${y_prediccion:.2f}k")
```
