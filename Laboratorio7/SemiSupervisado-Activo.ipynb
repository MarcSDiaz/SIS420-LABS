```python
import numpy as np
from sklearn.model_selection import train_test_split

data = np.load("geometric_shapes_dataset.npz")

X_train_raw = data["X_train"]     
X_val_raw   = data["X_val"]        
X_test_raw  = data["X_test"]       

X_train_flat = X_train_raw.reshape(len(X_train_raw), -1)
X_val_flat   = X_val_raw.reshape(len(X_val_raw), -1)
X_test_flat  = X_test_raw.reshape(len(X_test_raw), -1)

X_train_full = np.vstack([X_train_flat, X_val_flat])

X_test = X_test_flat

X_train, X_val = train_test_split(X_train_full, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.cluster import MiniBatchKMeans

k = 20

minibatch_kmeans = MiniBatchKMeans(n_clusters=k, batch_size=256, random_state=42, n_init=10)
clusters = minibatch_kmeans.fit_predict(X_train)
```
